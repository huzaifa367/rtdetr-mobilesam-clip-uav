```python
from roboflow import Roboflow
from pathlib import Path

ROBOFLOW_API_KEY = "qtnFdbzEhlnqVXaWWOt5"
WORKSPACE_ID = "hybridtunedmodeltraining"      # e.g. "my-workspace"
PROJECT_ID   = "regional_road"        # e.g. "uav-cars"
VERSION_NUM  = 3                     # your dataset version number

rf = Roboflow(api_key=ROBOFLOW_API_KEY)
project = rf.workspace(WORKSPACE_ID).project(PROJECT_ID)
dataset = project.version(VERSION_NUM).download("yolov8")  # creates a local folder

DATASET_DIR = Path(dataset.location)  # path to downloaded dataset folder
print("Downloaded to:", DATASET_DIR)
```

```python
from ultralytics import RTDETR
from pathlib import Path
import os, shutil

# Load model
model = RTDETR("/kaggle/working/runs/detect/HybridTunedModelForRelabeledImages/MobileSam_Clip/MobileSam_Clip/weights/best.pt")

src_images = Path("/kaggle/working/Regional_Road-3/train/images")

# Output structure for Roboflow
out_root = Path("/kaggle/working/roboflow_yolov8")
img_out  = out_root / "images"
lbl_out  = out_root / "labels"
img_out.mkdir(parents=True, exist_ok=True)
lbl_out.mkdir(parents=True, exist_ok=True)

def link_or_copy(src: Path, dst: Path):
    """Hardlink to avoid duplicating GBs on disk; fall back to copy if needed."""
    try:
        if dst.exists():
            return
        os.link(src, dst)   # no extra disk space if same filesystem
    except OSError:
        shutil.copy2(src, dst)

def unique_stem(p: Path) -> str:
    # If your filenames are unique already, this is fine.
    # If not, use parent folders to avoid collisions.
    # Example: "seq1__frame0001"
    rel = p.relative_to(src_images)
    stem = "__".join(rel.with_suffix("").parts)
    return stem

# Stream predictions so RAM stays stable
pred_iter = model.predict(
    source=str(src_images),
    imgsz=640,
    conf=0.50,   # up from 0.35
    iou=0.40,    # more aggressive suppression
    stream=True,      # IMPORTANT: no giant list in RAM
    save=False,
    save_txt=False,
    batch=4,          # adjust: 1/2/4/8
    workers=0,        # reduces RAM prefetching
    verbose=False
)

for r in pred_iter:
    p = Path(r.path)
    stem = unique_stem(p)

    # 1) Put image into output folder (hardlink/copy)
    # Keep original extension (.jpg/.png/...)
    img_dst = img_out / f"{stem}{p.suffix.lower()}"
    link_or_copy(p, img_dst)

    # 2) Write YOLOv8 label file (cls xc yc w h) normalized, NO confidence
    txt_path = lbl_out / f"{stem}.txt"

    if r.boxes is None or len(r.boxes) == 0:
        txt_path.write_text("")  # empty file is OK
        continue

    cls = r.boxes.cls.detach().cpu().numpy().astype(int)
    xywhn = r.boxes.xywhn.detach().cpu().numpy()  # normalized

    lines = [
        f"{c} {x:.6f} {y:.6f} {w:.6f} {h:.6f}"
        for c, (x, y, w, h) in zip(cls, xywhn)
    ]
    txt_path.write_text("\n".join(lines) + "\n")





print("Folder created.")
```
