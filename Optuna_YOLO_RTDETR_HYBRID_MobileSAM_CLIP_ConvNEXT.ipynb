```python
import sys; sys.path.insert(0, "/content")
from optuna_tune import tune_model, summarize

DATA = "splitted_dataset/data.yaml"   # normal yaml: train=train, val=val
ROOT = "/content/drive/MyDrive/MS_Passau_Thesis/Newly Modified Hybrid Encoder/results_optuna/optuna"
common = dict(data_yaml=DATA, save_root=ROOT, n_trials=30, epochs=10, fraction=0.2)
```
